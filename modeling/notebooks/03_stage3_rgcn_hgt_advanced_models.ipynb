# Stage 3 — R-GCN / HGT advanced models

This notebook trains the Stage 3 heterogeneous graph models when a PyG `HeteroData` export is available under `stage3_heterogeneous_gnn/pyg_export/heterodata.pt`. R-GCN is the recommended default because it handles typed PRING relations directly after flattening the heterogeneous graph into a multi-relational graph.


```python
from pathlib import Path

MODEL_INPUT = Path("/runs/current")
RGCN_OUTPUT = Path("/models/notebook_stage3_rgcn")
HGT_OUTPUT = Path("/models/notebook_stage3_hgt")
```

```python
from pring_modeling.stage3_rgcn import build_parser, run

args = build_parser().parse_args([
    "--modeling-dir", str(MODEL_INPUT),
    "--output-dir", str(RGCN_OUTPUT),
    "--epochs", "50",
    "--hidden-dim", "128",
    "--num-layers", "2",
    # "--score-candidates",  # uncomment when a candidate-pair file exists and memory allows
])
summary = run(args)
summary
```
